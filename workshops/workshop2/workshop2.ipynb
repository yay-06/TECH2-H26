# Solution

### Part 1 


````python
import pandas as pd

DATA_PATH = '../../Data'

df = pd.read_csv(f'{DATA_PATH}/titanic.csv')

df.head()
````

### Part 2

````python
print({df['Age'].isna().sum()})

df.info()
````

### Part 3

````python
# Part 3
avg_age = df['Age'].mean()

print(f'Average age : {avg_age}')
````

### Part 3

````python
import numpy as np 

print(f'Average age according to Numpy: {np.mean(df['Age'].to_numpy())}')
print(f'Average age acccording to Numpy: {np.nanmean(df['Age'].to_numpy())}')
````

### Part 4

````python
avg_age = np.round(avg_age)

df['Age'] = df['Age'].fillna(avg_age).astype(int)

df['Age']
````
